# Exploração inicial das bases

Inspeciona os esquemas e algumas linhas das bases CNEFE e de busca antes da normalização.


```python
# Objetivo: Abre as bases CNEFE e de busca em modo lazy e mostra os esquemas das colunas para orientar a exploracao.

# Importa Path para montar os caminhos e Polars para ler os CSVs.
from pathlib import Path
import polars as pl

# Define onde estao as bases brutas usadas na exploracao.
DATA_DIR = Path("../data/raw")

# Cria uma leitura lazy da base CNEFE, mantendo os campos como texto.
base = pl.scan_csv(DATA_DIR / "29_BA.csv", separator=";", infer_schema=False)

# Cria uma leitura lazy da base de endereços a localizar.
busca = pl.scan_csv(DATA_DIR / "2026_teste_tecnico_relinke_cidacs(in).csv", separator=";", infer_schema=False)

# Exibe os esquemas inferidos das duas fontes.
print("CNEFE:", base.collect_schema())
print("\nBusca:", busca.collect_schema())
```
